```python
print("__________________________""\n")
print("Controle de Vendas Mensais")
print("Digite 'FIM' no nome do produto para encerrar o cadastro\n")

with open('vendas.txt', 'w') as arquivo:
    while True:
        nome = input("Nome do produto: ")
        if nome.lower() == 'fim':
            break
        quantidade = int(input("Quantidade vendida: "))
        valor_unitario = float(input("Valor unitário: "))
        arquivo.write(f"{nome};{quantidade};{valor_unitario}\n")

with open('vendas.txt', 'r') as arquivo:
    linhas = arquivo.readlines()

vendas = []
for linha in linhas:
    nome, quantidade, valor_unitario = linha.strip().split(';')
    vendas.append({
        'nome': nome,
        'quantidade': int(quantidade),
        'valor_unitario': float(valor_unitario)
    })

faturamento_total = sum(v['quantidade'] * v['valor_unitario'] for v in vendas)
mais_vendido = max(vendas, key=lambda v: v['quantidade'])

print("\nResultados:")
print(f"Faturamento total do mês: R$ {faturamento_total:.2f}")
print(f"Produto mais vendido: {mais_vendido['nome']} - {mais_vendido['quantidade']} unidades")
```
