```python
def nome(nome,idade):
  resultado = f"Ola, meu nome é {nome} e tenho {idade} anos"
  return resultado

nome_pessoa = "Joao"
idade_pessoa = 18

resultado = nome(nome_pessoa, idade_pessoa)
print(resultado)
```

```python
def conversor(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit


celsius = float(input("digite a temperatura em celsius a ser convertida: "))
resultado = conversor(celsius)
print(resultado,"F°")
```

```python
def pares(inteiro):
  while inteiro > 0:
    if inteiro % 2 == 0:
      print(inteiro)
    inteiro -= 1
  return inteiro

numero = int(input("digite o numero: "))
resultado = pares(numero)
print(resultado)
```

```python
def media(nota1,nota2,nota3):
   media = (nota1 + nota2 + nota3) / 3
   return media

nota1 = float(input("digite a primeira nota: "))
nota2 = float(input("digite a segunda nota: "))
nota3 = float(input("digite a terceira nota: "))

resultado = media(nota1,nota2,nota3)
resultado2 = f"A média das três notas é igual à : {resultado}"
print(resultado2)
```
