```python
arquivo = open("notas_alunos.txt", "w")

for i in range(5):
    nome = input(f"Digite o nome do aluno {i+1}: ")
    nota = float(input(f"Digite a nota do aluno {i+1}: "))
    arquivo.write(f"{nome};{nota}\n")

arquivo.close()
```

```python
arquivo = open("notas_alunos.txt", "r")
conteudo = arquivo.read()
arquivo.close()

print("Conteúdo do arquivo notas_alunos.txt:")
print(conteudo)
```

```python

lista_alunos = []

with open("notas_alunos.txt", "r") as arquivo:
    for linha in arquivo:
        nome, nota_str = linha.strip().split(';')
        lista_alunos.append([nome, float(nota_str)])

soma_notas = 0
maior_nota = -1
aluno_maior_nota = ""

for aluno in lista_alunos:
    nome = aluno[0]
    nota = aluno[1]

    soma_notas += nota

    if nota > maior_nota:
        maior_nota = nota
        aluno_maior_nota = nome

if len(lista_alunos) > 0:
    media_notas = soma_notas / len(lista_alunos)
    print(f"A média da turma é: {media_notas:.2f}")
    print(f"O aluno com a maior nota é: {aluno_maior_nota} com nota {maior_nota:.2f}")
```
