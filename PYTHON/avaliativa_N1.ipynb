```python
def maior_nome(lista_de_nomes):
  maior = lista_de_nomes[0]
  for nome in lista_de_nomes:
    if len(nome) > len(maior):
      maior = nome
  return maior


nomes = [input("digite o primeiro nome: "),input("digite o segundo nome: "),input("digite o terceiro nome: ")]
nome_mais_longo = maior_nome(nomes)
print("O nome com o maior comprimento é:", nome_mais_longo)
```

```python
numeros = [1,2,3,4,5,6,7,8]
soma_pares = 0

for numero in numeros:
  if numero % 2 == 0:
    soma_pares += numero

print("A soma dos números pares é:", soma_pares)
```

```python
def inverter_lista(lista):
  return lista[::-1]

minha_lista = [1, 2, 3, 4, 5]
lista_invertida = inverter_lista(minha_lista)
print(lista_invertida)
```

```python
minha_lista = [1, 5, 2, 8, 5, 3, 2]

lista_sem_repetidos = list(set(minha_lista))

print(lista_sem_repetidos)
```

```python
lista = [120, 9, 902, 2, 1, 702, 99]
quantidade = 0
for numero in lista:
  if numero >= 100:
    quantidade += 1
    print("numeros maiores que 100: ", numero)

print("quantidade de numeros maiores que 100: ",quantidade)
```

```python
nomes = ["joao", "sueli", "Charlie brown", "chorao", "pedro sampaio"]
nome_busca = input("Digite um nome para procurar na lista: ")

if nome_busca in nomes:
  posicao = nomes.index(nome_busca)
  print(f"O nome '{nome_busca}' está na posição {posicao}.")
else:
  print(f"O nome '{nome_busca}' não foi encontrado na lista.")
```
