```python
from collections import Counter
paragrafo = """Neste dia ensolarado, o sol brilhava forte, o sol aquecia a terra. As aves cantavam no céu azul, o céu era vasto e azul. A brisa suave balançava as árvores, as árvores dançavam ao vento."""
arquivo = open("texto.txt", "w")
arquivo.write(paragrafo)
arquivo.close()

arquivo = open("texto.txt", "r")
conteudo = arquivo.read()
print("Frase selecionada:")
print(conteudo)
print("__________________________""\n")
arquivo.close()

texto_limpo = conteudo.lower()
for caractere in ".,!?;:":
    texto_limpo = texto_limpo.replace(caractere, "")

palavras = texto_limpo.split()
palavras_unicas = set(palavras)
total_palavras_diferentes = len(palavras_unicas)
contagem_palavras = Counter(palavras)
palavra_mais_repetida = contagem_palavras.most_common(1)[0][0]
quantidade_mais_repetida = contagem_palavras.most_common(1)[0][1]

resultado = f"{total_palavras_diferentes};{palavra_mais_repetida};{quantidade_mais_repetida}"

arquivo = open("analise.txt", "w")
arquivo.write(resultado)
arquivo.close()


print(f"Total de palavras diferentes: {total_palavras_diferentes}")
print(f"Palavra mais repetida: '{palavra_mais_repetida}'")
print(f"Quantidade de repetições: {quantidade_mais_repetida}")
print("__________________________""\n")

arquivo = open("analise.txt", "r")
conteudo_analise = arquivo.read()
print("Conteúdo do arquivo analise.txt:")
print(conteudo_analise)
arquivo.close()
```
