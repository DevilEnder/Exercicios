```python
print("                      \n")
print("Controle de estoque")
print("Digite 'FIM' no nome para encerrar o cadastro de produtos\n")

with open('estoque.txt', 'w') as arquivo:
    while True:
        nome = input("Nome do produto: ")
        if nome.lower() == 'fim':
            break
        preco = float(input("Preço: "))
        arquivo.write(f"{nome};{preco}\n")

with open('estoque.txt', 'r') as arquivo:
    linhas = arquivo.readlines()

produtos = []
for linha in linhas:
    nome, preco = linha.strip().split(';')
    produtos.append({'nome': nome, 'preco': float(preco)})

preco_medio = sum(p['preco'] for p in produtos) / len(produtos)
mais_caro = max(produtos, key=lambda p: p['preco'])

print("\nResultados:")
print(f"Preço médio: R$ {preco_medio:.2f}")
print(f"Produto mais caro: {mais_caro['nome']} - R$ {mais_caro['preco']:.2f}")
```
