```python
def fibonacci(numero):
  if numero <= 1:
    return numero
  return fibonacci(numero-1) + fibonacci(numero-2)


numero = int(input("informe o numero da sequencia que voce quer: "))
i = 0

for i in range(numero+1):
  print(fibonacci(i), end=", ")
```

```python
numero = int(input("informe o numero a fatorar: "))
if numero == 0:
  fatorial = 1
else:
  fatorial = numero
  contador = numero - 1
  while contador >= 1:
    fatorial *= contador
    contador -= 1

print(fatorial)
```
